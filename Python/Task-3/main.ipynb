```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Elevage.csv", delimiter=";")
df.head()
```

```python
df = df[(df.Produit.isin(["Bovins", "Ovins", "Porcins", "Poulets", "Canards", "Dindes"])) & (df.Unite=='1000 tetes') & (df.Annee!=2011)]
df.head()
```

```python
Produits=df.Produit.unique()
#plt.subplot(Produits.shape[0],1, i+1)
plt.figure(figsize=(12,8))
for i,produit in enumerate(Produits):
    df_grouped=df[df.Produit==produit].groupby(["Zone", "Annee"]).Valeur.sum().unstack()
    df_grouped.plot(kind='bar',stacked=False, figsize=(12,5))
    plt.title(f"{produit}")
```
